```python
with open("litanLogs.txt", "r") as f:
    s = f.read()
lines = s.split('\n')
# lines.append("0 0 0 0 0")

group = []
alldata = []

nowTime = 0

for line in lines:
    if line == "":
        continue
    content = line.split(' ')
    time = int(content[0])
    x = int(content[1])
    y = int(content[2])
    z = int(content[3])
    
    group.append([time, x, y, z])
    nowTime = time
    
    
```

```python
center = [23443, 19422]

plotCenter = [400, 400]

def getPos(x, y):
    X = (x - center[0]) / 64 * 10 + plotCenter[0]
    Y = -(y - center[1]) / 64 * 10 + plotCenter[1]
    return X, Y

def getVarPos(x, y, theta, r):
    Theta = theta / 256 * 2 * math.pi
    X = x - r * math.sin(Theta)
    Y = y - r * math.cos(Theta)
    return X, Y

def drawCircle(x, y, r, draw, outline, fill):
    X, Y = getPos(x, y)
    draw.ellipse([X - r, Y - r, X + r, Y + r], outline=outline, fill=fill)
    
def drawCirclePos(x, y, r, draw, outline, fill):
    draw.ellipse([x - r, y - r, x + r, y + r], outline=outline, fill=fill)

def drawText(x, y, text, fill):
    X, Y = getPos(x, y)
    draw.text((X, Y), text, fill=fill)
```
